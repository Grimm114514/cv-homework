```python
"""
SIFT 特征点匹配
完善1：修改提取参数, 填充ratio test
"""

import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import normalize

# 加载两幅图像
image1_path = './1.jpg'  # 替换为你的第一张图像路径
image2_path = './2.jpg'  # 替换为你的第二张图像路径

image1 = cv2.imread(image1_path, cv2.IMREAD_GRAYSCALE)
image2 = cv2.imread(image2_path, cv2.IMREAD_GRAYSCALE)


# 1. 初始化 SIFT 特征检测器
sift = cv2.SIFT_create()

# 2. 检测并计算描述符
keypoints1, descriptors1 = sift.detectAndCompute(image1, None)
keypoints2, descriptors2 = sift.detectAndCompute(image2, None)



# 3. 对描述符进行 L2 归一化
descriptors1 = normalize(descriptors1, norm='l2')
descriptors2 = normalize(descriptors2, norm='l2')

# 4. 使用 Brute-Force Matcher 进行特征点匹配
bf = cv2.BFMatcher(cv2.NORM_L2)
matches_nn = bf.match(descriptors1, descriptors2)
matches = bf.knnMatch(descriptors1, descriptors2, k=2)


# 5. 使用 Ratio Test 过滤匹配点
good_matches_ratio = []
for m,n in matches:
    if m.distance < 0.745 * n.distance:
        good_matches_ratio.append(m)
# 7. 绘制特征点匹配结果
img_matches_nn = cv2.drawMatches(image1, keypoints1, image2, 
                                  keypoints2, matches_nn[:256], 
                                  None, 
                                  matchColor=(0, 0, 255),
                                  flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)

img_matches_ratio = cv2.drawMatches(image1, keypoints1, image2, 
                                  keypoints2, good_matches_ratio[:256], 
                                  None, 
                                  matchColor=(0, 0, 255),
                                  flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)



image1_sift = cv2.drawKeypoints(image1, 
                               keypoints1, 
                               None, 
                               color=(0, 255, 0), 
                               flags=0)
image2_sift = cv2.drawKeypoints(image2, 
                               keypoints2, 
                               None, 
                               color=(0, 255, 0),
                               flags=0)


# 7. 显示结果
plt.figure(figsize=(20, 10))

plt.subplot(1, 2, 1)
plt.imshow(image1_sift, cmap='gray')
plt.axis("off")
plt.title('First Image with Keypoints')



plt.figure(figsize=(20, 10))
plt.imshow(img_matches_nn)
plt.title('Original Nearest Neighbor Matches')
plt.axis('off')

plt.figure(figsize=(20, 10))
plt.imshow(img_matches_ratio)
plt.axis("off")
plt.title('SIFT Feature Points Matching with Ratio Test')



plt.show()
```
